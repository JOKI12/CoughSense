Ce notebook charge les métriques réelles obtenues sur le test set pour comparer les performances de classification du SVM, Random Forest, KNN et CNN.

```python
import json
import pandas as pd
from pathlib import Path

metrics_path = Path('../models/model_metrics.json')
with open(metrics_path, 'r') as f:
    metrics = json.load(f)

summary_rows = []
for model, m in metrics.items():
    summary_rows.append({
        'Modèle': model,
        'Accuracy': f"{m['accuracy']*100:.2f} %",
        'Précision': f"{m.get('precision_macro', 0):.4f}",
        'Rappel': f"{m.get('recall_macro', 0):.4f}",
        'F1-Score': f"{m.get('f1_macro', 0):.4f}",
        'ROC-AUC': f"{m.get('roc_auc', 0):.4f}",
        'Latence (ms)': f"{m.get('inference_time_ms_per_sample', 0):.3f} ms"
    })

df_compare = pd.DataFrame(summary_rows)
print(df_compare.to_string(index=False))
```
