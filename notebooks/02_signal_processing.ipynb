**Master S3 Traitement du Signal**

Ce notebook détaille les étapes de filtrage Butterworth, débruitage par soustraction spectrale, FFT, STFT, Mel-Spectrogramme et extraction des 13 MFCCs.

```python
import sys
sys.path.insert(0, '..')
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from src.audio_loader import load_and_standardize_audio
from src.preprocessing import preprocess_signal
from src.cough_detection import segment_cough
from src.signal_analysis import analyze_signal
from src.features import extract_features

# Recherche d'un fichier audio réel
sample_files = list(Path('../data/raw').glob('**/*cough*.wav'))
audio_path = sample_files[0]
print('Fichier audio sélectionné :', audio_path)

audio_data = load_and_standardize_audio(audio_path, target_sample_rate=16000)
prep = preprocess_signal(audio_data.signal, sample_rate=16000)
seg = segment_cough(prep.processed_signal, sample_rate=16000)
analysis = analyze_signal(seg.segmented_signal, sample_rate=16000)
features = extract_features(seg.segmented_signal, sample_rate=16000)

print('Durée analysée :', round(analysis.time_domain.duration_sec, 2), 's')
print('Fréquence dominante :', round(analysis.frequency_domain.dominant_frequency_hz, 1), 'Hz')
print('Centroïde spectral :', round(analysis.frequency_domain.spectral_centroid_hz, 1), 'Hz')
print('Total des descripteurs extraits :', len(features.feature_names))
```
