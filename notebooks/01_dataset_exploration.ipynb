Ce notebook explore les métadonnées officielles et la distribution des participants issus des bases de données **Coswara (IISc Bangalore)** et **COUGHVID (EPFL)**.

```python
import pandas as pd
import json
from pathlib import Path
import matplotlib.pyplot as plt

# Chargement du résumé consolidé
summary_path = Path('../data/processed/dataset_summary.json')
if summary_path.exists():
    with open(summary_path) as f:
        summary = json.load(f)
    print('Résumé du jeu de données :', json.dumps(summary, indent=2))
else:
    print('Veuillez d\'abord exécuter : python training/prepare_dataset.py')
```

```python
# Analyse des splits
train_df = pd.read_csv('../data/processed/features_train.csv')
test_df = pd.read_csv('../data/processed/features_test.csv')
print('Train shape:', train_df.shape)
print('Test shape:', test_df.shape)

# Vérification de l'étanchéité des participants (Anti-Data Leakage)
train_parts = set(train_df['participant_id'])
test_parts = set(test_df['participant_id'])
overlap = train_parts & test_parts
print('Nombre de participants partagés entre Train et Test :', len(overlap))
assert len(overlap) == 0, 'Data leakage détecté !'
```
